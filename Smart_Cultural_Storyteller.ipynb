## 1. Problem Definition & Objective

Speech disfluency, especially stuttering, affects an individual’s ability to
communicate clearly and confidently. In real-world settings such as classrooms,
presentations, and storytelling, this can limit participation and self-expression.

### Objective
The objective of this project is to:
- Detect stuttered and normal speech using audio signal processing and deep learning
- Correct disfluent speech and generate a fluent version of the intended message
- Enable inclusive and accessible storytelling using AI


## 2. Data Understanding & Preparation

### Dataset Source
The dataset consists of speech recordings categorized into:
- Stuttered speech samples
- Normal speech samples

The data was collected from publicly available speech datasets and
organized into class-wise folders.

## 3. Model / System Design

### AI Technique Used
- Machine Learning and Deep Learning (Hybrid approach)
- MFCC-based feature extraction
- Neural Network classifier

### System Pipeline
1. Audio recording or input speech
2. Feature extraction using MFCC
3. Speech classification (stuttered / normal)
4. Rule-based and model-based decision fusion
5. Text correction and speech synthesis

### Design Justification
MFCC features effectively capture speech characteristics.
A neural network is used for classification due to its ability
to model non-linear patterns in speech signals.

## 6. Ethical Considerations & Responsible AI

- The dataset may not represent all accents, age groups, or speech conditions
- The system is intended as an assistive tool, not a medical diagnostic solution
- User data is processed locally and not stored permanently
- Bias and misclassification risks are acknowledged and minimized
